## 計算兩點的距離

```python
import math     # 引用 math 程式庫
x1, y1 = 3, 1   # 定義點 1 的座標
x2, y2 = 6, 5   # 定義點 2 的座標

# 計算距離並存入 dist 變數
dist = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
# print('兩點之間的距離：', dist)
print(f'兩點之間的距離：{dist}')
```

籠子裡只有雞和兔，牠們一共有**35個頭、94隻腳**，請問籠子裡各有幾隻雞和兔？

```python
# 程式碼的說明請參閱第6章
from sympy import Symbol, solve  # 從sympy匯入Symbol和solve函式
x = Symbol('x')  # 代表雞的數量
y = Symbol('y')  # 代表兔子的數量

total_heads = 35  # 頭的總數
total_feet = 94    # 腳的總數

eq1 = x + y - total_heads    # 建立方程式1
eq2 = 2*x + 4*y - total_feet  # 建立方程式2
solution = solve((eq1, eq2), (x, y))  # 解方程式

print("雞的數量:", solution[x])
print("兔子的數量:", solution[y])
```
